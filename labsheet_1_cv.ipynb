Question 1

Build a tool that loads any image format (JPG/PNG/BMP) and displays its key properties: dimensions, number of channels, data type, and file size.

```python
import cv2
import os
import matplotlib.pyplot as plt

image_path = "cv (2).jpg"

# Read original image
img = cv2.imread(image_path)

if img is None:
    print("Image not found!")
else:
    # Save image in different formats
    cv2.imwrite("cv2_output.jpg", img)
    cv2.imwrite("cv2_output.png", img)
    cv2.imwrite("cv2_output.bmp", img)

    # List of images
    images = [
        ("JPG", "cv2_output.jpg"),
        ("PNG", "cv2_output.png"),
        ("BMP", "cv2_output.bmp")
    ]

    print("IMAGE PROPERTIES")
    print("================")

    # Display all images
    plt.figure(figsize=(15, 5))

    for i, (name, filename) in enumerate(images):

        image = cv2.imread(filename)

        height, width = image.shape[:2]
        channels = image.shape[2]
        file_size = os.path.getsize(filename)

        print("\n", name)
        print("----------------")
        print("Dimensions :", width, "x", height)
        print("Channels   :", channels)
        print("Data Type  :", image.dtype)
        print("File Size  :", file_size, "bytes")

        # Convert BGR to RGB
        image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        # Display
        plt.subplot(1, 3, i + 1)
        plt.imshow(image_rgb)
        plt.title(name)
        plt.axis("off")

    plt.tight_layout()
    plt.show()
```

Question 5

Create a batch image resizer that reads multiple images from a folder and saves resized copies while preserving aspect ratio.

```python
import cv2
import os
import matplotlib.pyplot as plt

# Your image name
image_path = "cv (2).jpg"

# Read image
img = cv2.imread(image_path)

if img is None:
    print("Image not found! Check the file name.")
else:
    # Maximum dimensions
    max_width = 500
    max_height = 500

    # Original dimensions
    h, w = img.shape[:2]

    # Calculate scale while preserving aspect ratio
    scale = min(max_width / w, max_height / h)

    new_width = int(w * scale)
    new_height = int(h * scale)

    # Resize image
    resized = cv2.resize(img, (new_width, new_height))

    # Save resized image
    cv2.imwrite("cv2_resized.jpg", resized)

    print("Original Size :", w, "x", h)
    print("Resized Size  :", new_width, "x", new_height)
    print("Resized image saved as cv2_resized.jpg")

    # Display images
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
    plt.title("Resized Image")
    plt.axis("off")

    plt.show()
```

Question 7

Build a tool that splits an image into its individual Red, Green and Blue channels, then reconstructs it and verifies pixel-level accuracy.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = "cv (2).jpg"

img = cv2.imread(image_path)

if img is None:
    print("Image not found!")
else:

    # OpenCV uses BGR order
    B, G, R = cv2.split(img)

    # Reconstruct
    reconstructed = cv2.merge([B, G, R])

    # Save reconstructed image
    cv2.imwrite("cv (2)_reconstructed.jpg", reconstructed)

    # Check pixel-level accuracy
    difference = cv2.absdiff(img, reconstructed)

    if np.max(difference) == 0:
        print("Reconstruction is pixel-level accurate.")
    else:
        print("Images are different.")

    # Display original
    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(reconstructed, cv2.COLOR_BGR2RGB))
    plt.title("Reconstructed Image")
    plt.axis("off")

    plt.show()

    # Display individual channels
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 3, 1)
    plt.imshow(R, cmap="gray")
    plt.title("Red Channel")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(G, cmap="gray")
    plt.title("Green Channel")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(B, cmap="gray")
    plt.title("Blue Channel")
    plt.axis("off")

    plt.show()
```

Question 9

Develop a program to rotate and flip an image by user-defined angles and directions, saving all transformed versions.

```python
import cv2
import matplotlib.pyplot as plt

image_path = "cv (2).jpg"

img = cv2.imread(image_path)

if img is None:
    print("Image not found!")
else:

    # Rotation angle
    angle = 45

    h, w = img.shape[:2]
    center = (w // 2, h // 2)

    matrix = cv2.getRotationMatrix2D(center, angle, 1)

    rotated = cv2.warpAffine(img, matrix, (w, h))

    # Flips
    horizontal = cv2.flip(img, 1)
    vertical = cv2.flip(img, 0)

    # Save images
    cv2.imwrite("cv (2)_rotated.jpg", rotated)
    cv2.imwrite("cv (2)_horizontal_flip.jpg", horizontal)
    cv2.imwrite("cv (2)_vertical_flip.jpg", vertical)

    # Display all
    plt.figure(figsize=(12, 8))

    plt.subplot(2, 2, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(2, 2, 2)
    plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
    plt.title("Rotated 45°")
    plt.axis("off")

    plt.subplot(2, 2, 3)
    plt.imshow(cv2.cvtColor(horizontal, cv2.COLOR_BGR2RGB))
    plt.title("Horizontal Flip")
    plt.axis("off")

    plt.subplot(2, 2, 4)
    plt.imshow(cv2.cvtColor(vertical, cv2.COLOR_BGR2RGB))
    plt.title("Vertical Flip")
    plt.axis("off")

    plt.show()
```

Question 10

Build a simple image metadata viewer that extracts EXIF data, if present, along with basic pixel statistics such as mean, minimum and maximum.

```python
from PIL import Image, ExifTags
import numpy as np
import matplotlib.pyplot as plt

image_path = "cv (2).jpg"

img = Image.open(image_path)

print("Image Metadata")
print("----------------")
print("Format     :", img.format)
print("Dimensions :", img.size)
print("Mode       :", img.mode)

# EXIF information
exif = img.getexif()

if exif:
    print("\nEXIF Data")
    print("---------")

    for tag_id, value in exif.items():
        tag = ExifTags.TAGS.get(tag_id, tag_id)
        print(tag, ":", value)
else:
    print("\nNo EXIF data found.")

# Pixel statistics
array = np.array(img)

print("\nPixel Statistics")
print("----------------")
print("Mean    :", np.mean(array))
print("Minimum :", np.min(array))
print("Maximum :", np.max(array))

# Display image
plt.imshow(img)
plt.title("Original Image - cv (2)")
plt.axis("off")
plt.show()
```
